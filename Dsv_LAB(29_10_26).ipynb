##SQLITE

```python
import sqlite3
conn=sqlite3.connect("students.db")
```

```python
cursor=conn.cursor()
print("Database Created Successfully")
```

```python
cursor.execute("""
    Create TABLE IF NOT EXISTS students_table(
    Name TEXT,
    ROLL_No INTEGER,
    Marks INTEGER)
    """
)
print("Table is created successfully")
```

```python
#write data into table
import sqlite3
conn = sqlite3.connect("students.db")
cursor = conn.cursor()

cursor.execute(
    "INSERT INTO students_table VALUES(?, ?, ?)",
    ("Amit", 101, 90)
)
cursor.execute(
    "INSERT INTO students_table VALUES('Aman', 102, 95)"
)
cursor.execute(
    "INSERT INTO students_table VALUES('Ayaansh', 123, 98)"
)
```

```python
conn.commit()
print("Records inserted successfully !!")
conn.close()
```

```python
#retrieve | read the data
conn=sqlite3.connect("students.db")
cursor=conn.cursor()
cursor.execute("SELECT * FROM students_table")
rows=cursor.fetchall()
for row in rows:
  print(row)
conn.close()
```
